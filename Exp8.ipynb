```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from ddgs import DDGS

print("Libraries imported successfully!")
```

```python
class ResearchState(TypedDict):
    topic: str
    research_data: str
    draft: str
    final_paper: str
```

```python
def research_agent(state: ResearchState):

    print("\n[Research Agent] Collecting information...")

    topic = state["topic"]

    try:
        results = DDGS().text(
            topic,
            max_results=5
        )

        research = ""

        for i, result in enumerate(results, 1):

            research += (
                f"\nSource {i}\n"
                f"Title: {result.get('title', '')}\n"
                f"Information: {result.get('body', '')}\n"
                f"URL: {result.get('href', '')}\n"
            )

        if not research:
            research = "No search results found."

    except Exception as e:
        research = f"Search failed: {e}"

    print("[Research Agent] Research completed!")

    return {"research_data": research}
```

```python
def writer_agent(state: ResearchState):

    print("\n[Writer Agent] Writing research paper...")

    topic = state["topic"]
    research = state["research_data"]

    draft = f"""
RESEARCH PAPER

Title: {topic}

1. ABSTRACT

This research paper discusses {topic}.
It examines the important concepts, applications,
benefits and challenges associated with this topic.

2. INTRODUCTION

{topic} is an important area of research in
modern technology. It has gained attention
because of its potential applications and impact.

3. RESEARCH FINDINGS

The following information was collected
by the Research Agent:

{research}

4. APPLICATIONS

The topic has several practical applications
in education, industry, business and technology.

5. ADVANTAGES

- Improves efficiency.
- Supports innovation.
- Provides useful solutions to real-world problems.

6. CHALLENGES

- Implementation difficulties.
- Cost and resource requirements.
- Security and reliability concerns.

7. CONCLUSION

The study highlights the importance of
{topic} and its potential for future development.

8. REFERENCES

The research sources are listed in the
research findings above.
"""

    print("[Writer Agent] Draft completed!")

    return {"draft": draft}
```

```python
def editor_agent(state: ResearchState):

    print("\n[Editor Agent] Reviewing research paper...")

    paper = state["draft"]

    # Improve formatting
    paper = paper.replace(
        "RESEARCH PAPER",
        "FINAL RESEARCH PAPER"
    )

    # Remove unnecessary blank lines
    lines = paper.splitlines()
    cleaned_lines = []

    for line in lines:
        if line.strip() or (
            cleaned_lines and cleaned_lines[-1].strip()
        ):
            cleaned_lines.append(line)

    final_paper = "\n".join(cleaned_lines)

    print("[Editor Agent] Editing completed!")

    return {"final_paper": final_paper}
```

```python
# Create the workflow
workflow = StateGraph(ResearchState)

# Add agents as nodes
workflow.add_node("Research_Agent", research_agent)
workflow.add_node("Writer_Agent", writer_agent)
workflow.add_node("Editor_Agent", editor_agent)

# Define execution order
workflow.add_edge(START, "Research_Agent")
workflow.add_edge("Research_Agent", "Writer_Agent")
workflow.add_edge("Writer_Agent", "Editor_Agent")
workflow.add_edge("Editor_Agent", END)

# Compile the workflow
app = workflow.compile()

print("Multi-Agent System Created Successfully!")
```

```python
# Get research topic from user
topic = input("Enter your research topic: ")

# Execute all agents
result = app.invoke({
    "topic": topic,
    "research_data": "",
    "draft": "",
    "final_paper": ""
})

# Display final output
print("\n===================================")
print("       FINAL RESEARCH PAPER")
print("===================================")

print(result["final_paper"])

# Save research paper as a text file
with open("research_paper.txt", "w", encoding="utf-8") as file:
    file.write(result["final_paper"])

print("\nResearch paper saved successfully!")
```
