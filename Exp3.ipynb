```python
import time


class SmartHomeEnvironment:
    def __init__(self, temperature=30):
        self.temperature = temperature
        self.ac_status = False

    def get_temperature(self):
        return self.temperature

    def set_ac(self, status):
        self.ac_status = status

    def update_temperature(self):
        if self.ac_status:
            self.temperature -= 1
        else:
            self.temperature += 0.5


class TemperatureAgent:
    def __init__(self, environment):
        self.environment = environment

    def perceive(self):
        return self.environment.get_temperature()

    def act(self, temperature):
        if temperature > 24:
            self.environment.set_ac(True)
            print("Temperature:", temperature, "°C")
            print("Action: AC turned ON")

        elif temperature < 22:
            self.environment.set_ac(False)
            print("Temperature:", temperature, "°C")
            print("Action: AC turned OFF")

        else:
            print("Temperature:", temperature, "°C")
            print("Action: Temperature is comfortable")
            print("AC Status:",
                  "ON" if self.environment.ac_status else "OFF")

        print("-" * 40)


environment = SmartHomeEnvironment(temperature=30)

agent = TemperatureAgent(environment)

for i in range(15):
    temperature = agent.perceive()
    agent.act(temperature)

    environment.update_temperature()

    time.sleep(1)
```
