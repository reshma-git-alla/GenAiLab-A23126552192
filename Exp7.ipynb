```python
import yfinance as yf
from ddgs import DDGS

print("Libraries imported successfully!")
```

```python
def calculate_emi(principal, annual_rate, years):

    monthly_rate = annual_rate / (12 * 100)
    months = years * 12

    if monthly_rate == 0:
        emi = principal / months
    else:
        emi = (
            principal * monthly_rate *
            (1 + monthly_rate) ** months
        ) / (
            (1 + monthly_rate) ** months - 1
        )

    total_payment = emi * months
    total_interest = total_payment - principal

    return emi, total_payment, total_interest


print("EMI Calculator Tool Created!")
```

```python
def get_stock_price(symbol):

    try:
        stock = yf.Ticker(symbol)
        history = stock.history(period="5d")

        if history.empty:
            return "Stock information not found."

        price = history["Close"].dropna().iloc[-1]

        return f"Latest available closing price of {symbol}: {price:.2f}"

    except Exception as e:
        return f"Error: {e}"


print("Stock Price Tool Created!")
```

```python
def search_financial_info(query):

    try:
        results = DDGS().text(
            query + " financial information",
            max_results=5
        )

        for i, result in enumerate(results, 1):

            print(f"\nResult {i}")
            print("Title:", result.get("title"))
            print("Information:", result.get("body"))
            print("Source:", result.get("href"))

    except Exception as e:
        print("Search Error:", e)


print("Financial Search Tool Created!")
```

```python
def financial_agent():

    print("================================")
    print("   AI FINANCIAL ASSISTANT")
    print("================================")

    print("1. EMI Calculator")
    print("2. Stock Price Retrieval")
    print("3. Financial Information Search")
    print("Type exit to stop.")

    while True:

        query = input("\nEnter your request: ").strip()

        if query.lower() == "exit":
            print("Thank you for using Financial Assistant!")
            break

        elif query.lower() in ["emi", "calculate emi", "loan emi"]:

            try:
                principal = float(input("Enter loan amount: "))
                rate = float(input("Enter annual interest rate: "))
                years = int(input("Enter loan duration in years: "))

                if principal <= 0 or rate < 0 or years <= 0:
                    print("Please enter valid values.")
                    continue

                emi, total, interest = calculate_emi(
                    principal, rate, years
                )

                print("\n--- EMI RESULT ---")
                print(f"Monthly EMI: Rs. {emi:,.2f}")
                print(f"Total Payment: Rs. {total:,.2f}")
                print(f"Total Interest: Rs. {interest:,.2f}")

            except ValueError:
                print("Please enter valid numeric values.")

        elif query.lower().startswith("stock"):

            symbol = query[5:].strip()

            if not symbol:
                symbol = input("Enter stock symbol: ")

            print(get_stock_price(symbol))

        elif query.lower().startswith("search"):

            topic = query[6:].strip()

            if not topic:
                topic = input("Enter financial topic: ")

            search_financial_info(topic)

        else:
            print("Invalid request!")
            print("Try EMI, STOCK RELIANCE.NS, or SEARCH mutual funds.")


financial_agent()
```
