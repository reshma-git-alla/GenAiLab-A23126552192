```python
import json
import os
from datetime import datetime

print("Libraries imported successfully!")
```

```python
from google.colab import drive

drive.mount('/content/drive')

MEMORY_FILE = "/content/drive/MyDrive/customer_memory.json"

# Create memory file if it does not exist
if not os.path.exists(MEMORY_FILE):
    with open(MEMORY_FILE, "w") as file:
        json.dump({}, file)

print("Long-term memory initialized!")
```

```python
# Short-term memory stores current conversation
short_term_memory = {}

# Load long-term memory from JSON file
def load_long_term_memory():

    with open(MEMORY_FILE, "r") as file:
        return json.load(file)


# Save long-term memory to JSON file
def save_long_term_memory(memory):

    with open(MEMORY_FILE, "w") as file:
        json.dump(memory, file, indent=4)


print("Short-term and long-term memory created!")
```

```python
def customer_support_agent(customer_id, message):

    memory = load_long_term_memory()

    # Create customer profile if new
    if customer_id not in memory:
        memory[customer_id] = {
            "complaints": [],
            "total_interactions": 0
        }

    customer = memory[customer_id]

    # Initialize short-term memory
    if customer_id not in short_term_memory:
        short_term_memory[customer_id] = []

    # Store current conversation
    short_term_memory[customer_id].append({
        "user": message,
        "time": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    })

    # Keep only the latest 10 messages
    short_term_memory[customer_id] = (
        short_term_memory[customer_id][-10:]
    )

    customer["total_interactions"] += 1

    # Identify complaint-related messages
    complaint_words = [
        "complaint", "problem", "issue", "not working",
        "damaged", "refund", "delay", "late", "broken"
    ]

    is_complaint = any(
        word in message.lower() for word in complaint_words
    )

    # Retrieve previous complaints
    previous_complaints = customer["complaints"]

    if is_complaint:

        complaint = {
            "message": message,
            "date": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "status": "Open"
        }

        customer["complaints"].append(complaint)

        response = (
            "We are sorry for the inconvenience. "
            "Your complaint has been recorded. "
            "Our support team can review this issue."
        )

        if len(previous_complaints) > 0:
            response += (
                "\nWe also found your previous complaint "
                "in our records. We will consider your "
                "earlier issue while assisting you."
            )

    else:

        if previous_complaints:

            last_complaint = previous_complaints[-1]["message"]

            response = (
                "Welcome back! We remember your previous issue: "
                + last_complaint
                + ". How can we help you further?"
            )

        else:

            response = (
                "Hello! Welcome to customer support. "
                "How can I assist you today?"
            )

    # Save updated long-term memory
    save_long_term_memory(memory)

    # Store agent response in short-term memory
    short_term_memory[customer_id][-1]["agent"] = response

    return response


print("Customer Support Agent created successfully!")
```

```python
def run_customer_support():

    print("\n====================================")
    print("     AI CUSTOMER SUPPORT SYSTEM")
    print("====================================")

    customer_id = input("Enter Customer ID: ")

    print("\nWelcome to Customer Support!")
    print("Type 'exit' to end the conversation.")

    while True:

        message = input("\nYou: ").strip()

        if message.lower() == "exit":
            print("Agent: Thank you for contacting us!")
            break

        if not message:
            print("Agent: Please enter a message.")
            continue

        response = customer_support_agent(
            customer_id, message
        )

        print("\nAgent:", response)


run_customer_support()
```
